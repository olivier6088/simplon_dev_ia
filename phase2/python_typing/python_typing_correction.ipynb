## 1. Rappels : types de base en Python

## 2. Annotations de type simples

Python permet d’**annoter** les variables et les fonctions pour indiquer les types attendus.

Exemples :

```python
x: int = 3
nom: str = "Alice"

def ajouter(a: int, b: int) -> int:
    return a + b
```

⚠️ Ces annotations ne sont pas vérifiées par Python à l’exécution :
elles servent aux outils (IDE, mypy, pyright) et à la documentation.

````python
# Correction 2.1 : annotations simples sur des variables

x: int = 3
titre: str = "Cours de Python"
est_actif: bool = True

print(x, titre, est_actif)
````

### 2.2 – Annotations de fonction

On peut annoter :
- les paramètres
- le type de retour avec `-> ...`.

Exemple :

```python
def ajouter(a: int, b: int) -> int:
    return a + b
```

````python
# Correction 2.2 : fonction annotée

from typing import Any


def dire_bonjour(nom: str) -> None:
    print(f"Bonjour {nom} !")


dire_bonjour("Alice")
# À l'exécution, ceci fonctionne mais un type checker le signalerait comme erreur :
# dire_bonjour(123)
````

## 3. Collections typées & syntaxes modernes/anciennes

Pour les collections, on peut préciser le type des éléments.

### Syntaxe moderne (Python 3.9+)

- `list[int]`
- `dict[str, float]`
- `set[str]`
- `tuple[int, str]`

### Syntaxe ancienne (versions < 3.9)

- `List[int]` (avec `from typing import List`)
- `Dict[str, float]` (avec `from typing import Dict`)

Les deux syntaxes signifient la même chose pour un type checker.

````python
# Correction 3.1 : listes typées

nombres: list[int] = [1, 2, 3]
noms: list[str] = ["Alice", "Bob", "Charlie"]

print(nombres)
print(noms)
````

````python
# Correction 3.2 : dictionnaire typé

notes: dict[str, float] = {
    "Alice": 15.5,
    "Bob": 12.0,
}
print(notes)
````

````python
# Correction 3.3 : fonction avec list[str]

def compter_noms(noms: list[str]) -> int:
    return len(noms)


print(compter_noms(["Alice", "Bob"]))
print(compter_noms([]))
````

## 4. Types avancés : `Optional`, `Union`, `Any`, `Literal`

On utilise le module `typing` pour exprimer des types plus riches.

- `Optional[T]` signifie : `T` ou `None`.
- `Union[A, B]` signifie : A **ou** B.
- `Any` désactive le typage statique.
- `Literal[...]` restreint à quelques valeurs précises.

````python
# Correction 4.1 : utilisation de Optional
from typing import Optional


def dire_bonjour_optional(prenom: Optional[str] = None) -> None:
    if prenom is None:
        print("Bonjour le monde !")
    else:
        print(f"Bonjour {prenom} !")


dire_bonjour_optional("Alice")
dire_bonjour_optional(None)
````

````python
# Correction 4.2 : utilisation de Union
from typing import Union

Nombre = Union[int, float]


def doubler(x: Nombre) -> Nombre:
    return 2 * x


print(doubler(10))
print(doubler(3.14))
````

````python
# Correction 4.3 : Any accepte tous les types
from typing import Any

valeur: Any = 42
print(valeur, type(valeur))

valeur = "texte"
print(valeur, type(valeur))

valeur = [1, 2, 3]
print(valeur, type(valeur))
````

````python
# Correction 4.4 : Literal
from typing import Literal

Couleur = Literal["rouge", "vert", "bleu"]


def afficher_couleur(c: Couleur) -> None:
    print(f"Couleur choisie : {c}")


afficher_couleur("rouge")
# Un type checker se plaindrait de cet appel :
# afficher_couleur("jaune")
````

## 5. Classes comme types

Une classe peut aussi être utilisée comme type dans les annotations.

```python
class User:
    def __init__(self, name: str, age: int) -> None:
        self.name = name
        self.age = age

def show_user(user: User) -> None:
    print(user.name, user.age)
```

````python
# Correction 5.1 : classe User et annotation de type

class User:
    def __init__(self, name: str, age: int) -> None:
        self.name = name
        self.age = age


def show_user(user: User) -> None:
    print(f"Utilisateur : {user.name} ({user.age} ans)")


u = User("Alice", 30)
show_user(u)
````

````python
# Correction 5.2 : classe Product et fonction utilisant ce type

class Product:
    def __init__(self, name: str, price: float) -> None:
        self.name = name
        self.price = price


def print_product(p: Product) -> None:
    print(f"Produit : {p.name} - Prix : {p.price} €")


p = Product("Stylo", 1.5)
print_product(p)
````

## 6. Alias de types et `Annotated`

On peut créer des **alias de types** pour rendre le code plus lisible :

```python
UserId = int
Score = float
```

Avec `Annotated`, on peut associer des **métadonnées** à un type, ce qui est utilisé par des frameworks comme FastAPI.

```python
from typing import Annotated

UserId = Annotated[int, "Identifiant d'utilisateur"]
```

````python
# Correction 6.1 : alias de types

UserId = int
Price = float


def display_price(p: Price) -> None:
    print(f"Prix : {p} €")


display_price(9.99)
````

````python
# Correction 6.2 : Annotated pour ajouter des métadonnées
from typing import Annotated

UserId = Annotated[int, "Identifiant d'utilisateur"]
EmailStr = Annotated[str, "Adresse email"]


def send_email(to: EmailStr, subject: str, body: str) -> None:
    print(f"Envoi d'un email à {to} : {subject}\n{body}")


send_email("test@example.com", "Sujet", "Corps du message")
````

## 7. Fonctions génériques avec `TypeVar`

Un `TypeVar` permet de définir une fonction générique qui fonctionne avec différents types,
tout en conservant l'information de type.

```python
from typing import TypeVar

T = TypeVar("T")

def identite(x: T) -> T:
    return x
```

````python
# Correction 7.1 : fonction générique identite
from typing import TypeVar

T = TypeVar("T")


def identite(x: T) -> T:
    return x


print(identite(42))
print(identite("hello"))
print(identite([1, 2, 3]))
````

````python
# Correction 7.2 : dernier_element générique
from typing import TypeVar

U = TypeVar("U")


def dernier_element(items: list[U]) -> U | None:
    if not items:
        return None
    return items[-1]


print(dernier_element([1, 2, 3]))
print(dernier_element(["a", "b"]))
print(dernier_element([]))
````

## 8. Modèles Pydantic comme types (lien avec FastAPI)

FastAPI utilise **Pydantic** pour définir des modèles de données **typés** et **validés**.

```python
from pydantic import BaseModel

class UserModel(BaseModel):
    id: int
    name: str
    is_active: bool = True
```

Ces modèles peuvent être utilisés comme types dans les annotations de fonctions.

````python
# Correction 8.1 : modèle Pydantic et utilisation comme type
try:
    from pydantic import BaseModel
except ImportError:
    BaseModel = None

if BaseModel is not None:

    class UserModel(BaseModel):
        id: int
        name: str
        is_active: bool = True


    def deactivate(user: "UserModel") -> "UserModel":
        user.is_active = False
        return user


    u = UserModel(id=1, name="Alice")
    print(u)
    u = deactivate(u)
    print(u)
else:
    print("Pydantic n'est pas installé dans cet environnement.")
````

## 9. Motivation : pourquoi typer ? et outils de vérification

Exemple inspiré de la doc FastAPI :

```python
def get_full_name(first_name, last_name):
    full_name = first_name.title() + " " + last_name.title()
    return full_name
```

Cette fonction marche sans types, mais :
- l'éditeur ne sait pas quels sont les types
- un appel incorrect ne sera pas repéré avant l'exécution.

Avec des annotations :

```python
def get_full_name(first_name: str, last_name: str) -> str:
    ...
```

Un type checker peut alors signaler les erreurs.

````python
# Correction 9.1 : ajout d'annotations et appels


def get_full_name(first_name: str, last_name: str) -> str:
    full_name = first_name.title() + " " + last_name.title()
    return full_name


print(get_full_name("john", "doe"))
# À l'exécution, ceci planterait probablement car int n'a pas de .title():
# get_full_name(123, "doe")
````

### 9.2 – Outils : `mypy` et `pyright`

Les annotations seules ne suffisent pas : il faut un outil pour les vérifier.

- Installation de mypy :

```bash
pip install mypy
mypy mon_fichier.py
```

- Installation de pyright :

```bash
npm install -g pyright
pyright mon_fichier.py
```

Ces outils signalent les incohérences de types **avant l'exécution**.
