## Partie A - Pandas

```python
import pandas as pd
```

```python
# 1. Charger le fichier `customers.csv`
customers = pd.read_csv('../../data/customers.csv')

customers.head()
```

```python
# 2. Afficher uniquement la colonne `name`.
customers['name']
```

```python
# 3. Lister les clients de Paris.
customers[customers['city'] == 'Paris']
```

```python
# 4. Trier les clients par âge décroissant.  
customers.sort_values(by='age', ascending=False)
```

```python
# 5. Compter combien de clients viennent de chaque ville.  
customers['city'].value_counts()
```

```python
# 6. Sauvegarder les clients de Lyon dans `data/lyon_clients.csv`.  
customers[customers['city'] == 'Lyon'].to_csv('../../data/lyon_clients.csv', index=False)
```

```python
# *(Optionnel)* Lister les clients dont le nom commence par “A”.  
customers[customers['name'].str.startswith('A')]
```
