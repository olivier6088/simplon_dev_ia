## Partie B - Pandas et SQL

```python
import pandas as pd
import sqlite3
con = sqlite3.connect('../../data/marketing.db')
```

```python
# 7. Charger la table `orders` depuis la base `marketing.db`
orders = pd.read_sql('SELECT * FROM orders', con)
```

```python
# 8. Lister les 5 premières commandes avec le nom du client
orders_with_customers = pd.read_sql('''
    SELECT o.*, c.name AS customer_name
    FROM orders o
    JOIN customers c ON o.customer_id = c.id
    LIMIT 5
''', con)

orders_with_customers
```

```python
# 9. Calculer le total dépensé par chaque client
total_per_customer = pd.read_sql('''
    SELECT c.id AS customer_id, c.name AS customer_name, SUM(o.amount) AS total_spent
    FROM customers c
    JOIN orders o ON c.id = o.customer_id
    GROUP BY c.id, c.name
''', con)

total_per_customer
```

```python
# 10. Calculer le panier moyen par ville.
average_basket_per_city = pd.read_sql('''
    SELECT c.city, AVG(o.amount) AS average_basket
    FROM customers c
    JOIN orders o ON c.id = o.customer_id
    GROUP BY c.city
''', con)

average_basket_per_city
```

```python
# 11. Exporter le rapport en CSV
average_basket_per_city.to_csv('average_basket_per_city.csv', index=False)
```

```python
# *(Optionnel)* Trouver le top 3 clients par montant total.
top_3_customers = total_per_customer.sort_values(by='total_spent', ascending=False).head(3)
top_3_customers
```
